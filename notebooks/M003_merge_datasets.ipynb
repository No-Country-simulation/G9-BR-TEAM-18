```python
import pandas as pd
from pathlib import Path
```

```python
BASE_DIR = Path("../data/processed/for_ml")

ROTULED_PATH = BASE_DIR / "rotuled_ml_processed.csv"

PPH_PATH = Path("../ml-service/data/pph-data-complete.csv")
```

```python
rotuled = pd.read_csv(ROTULED_PATH)

pph = pd.read_csv(PPH_PATH)

print(rotuled.shape)

print(pph.shape)
```

```python
print("Colunas do rotuled:")
print(rotuled.columns.tolist())

print("\nColunas do pph:")
print(pph.columns.tolist())
```

```python
entrevistas_rotuled = (
    rotuled["entrevista"]
    .reset_index(drop=True)
)

entrevistas_pph = (
    pph["ENTREVISTA"]
    .reset_index(drop=True)
)

entrevistas_alinhadas = entrevistas_rotuled.equals(entrevistas_pph)

print("Sequência de ENTREVISTA alinhada:", entrevistas_alinhadas)
```

```python
comparacao_identificacao = pd.DataFrame({
    "entrevista_rotuled": rotuled["entrevista"].reset_index(drop=True),
    "entrevista_pph": pph["ENTREVISTA"].reset_index(drop=True),
    "regiao_rotuled": rotuled["regiao"].reset_index(drop=True),
    "regiao_pph": pph["REGIAO"].reset_index(drop=True),
    "uf_rotuled": rotuled["uf"].reset_index(drop=True),
    "uf_pph": pph["UF"].reset_index(drop=True),
    "municipio_rotuled": rotuled["municipio"].reset_index(drop=True),
    "municipio_pph": pph["MUNICIPIO"].reset_index(drop=True),
})

comparacao_identificacao["entrevista_igual"] = (
    comparacao_identificacao["entrevista_rotuled"]
    == comparacao_identificacao["entrevista_pph"]
)

comparacao_identificacao["regiao_igual"] = (
    comparacao_identificacao["regiao_rotuled"]
    == comparacao_identificacao["regiao_pph"]
)

comparacao_identificacao["uf_igual"] = (
    comparacao_identificacao["uf_rotuled"]
    == comparacao_identificacao["uf_pph"]
)

comparacao_identificacao["municipio_igual"] = (
    comparacao_identificacao["municipio_rotuled"]
    == comparacao_identificacao["municipio_pph"]
)

print(
    "ENTREVISTA iguais:",
    comparacao_identificacao["entrevista_igual"].sum(),
    "de",
    len(comparacao_identificacao)
)

print(
    "REGIAO iguais:",
    comparacao_identificacao["regiao_igual"].sum(),
    "de",
    len(comparacao_identificacao)
)

print(
    "UF iguais:",
    comparacao_identificacao["uf_igual"].sum(),
    "de",
    len(comparacao_identificacao)
)

print(
    "MUNICIPIO iguais:",
    comparacao_identificacao["municipio_igual"].sum(),
    "de",
    len(comparacao_identificacao)
)
```

```python
colunas_pph = [
    "ENTREVISTA",
    "REGIAO",
    "UF",
    "MUNICIPIO",
]

colunas_pph += [
    coluna
    for coluna in pph.columns
    if coluna.startswith("qtd_")
]

colunas_pph += [
    coluna
    for coluna in pph.columns
    if coluna.startswith("kwh_categoria_")
]

colunas_pph += [
    "consumo_real_medio_kwh_mes"
]

colunas_pph += [
    coluna
    for coluna in pph.columns
    if coluna.startswith("habito_")
]

colunas_rotuled = [
    "produtos_maior_consumo",
    "categoria_maior_consumo",
]

print("Quantidade de colunas selecionadas do PPH:", len(colunas_pph))
print("Quantidade de colunas selecionadas do rotuled:", len(colunas_rotuled))

print("\nColunas do PPH:")
print(colunas_pph)

print("\nColunas do rotuled:")
print(colunas_rotuled)
```

```python
if len(rotuled) != len(pph):
    raise ValueError(
        "Os datasets possuem quantidades diferentes de linhas. "
        "A concatenação foi interrompida."
    )

if not rotuled["entrevista"].reset_index(drop=True).equals(
    pph["ENTREVISTA"].reset_index(drop=True)
):
    raise ValueError(
        "A sequência da coluna ENTREVISTA está diferente. "
        "A concatenação foi interrompida."
    )

if not rotuled["regiao"].reset_index(drop=True).equals(
    pph["REGIAO"].reset_index(drop=True)
):
    raise ValueError(
        "A coluna REGIAO está desalinhada."
    )

if not rotuled["uf"].reset_index(drop=True).equals(
    pph["UF"].reset_index(drop=True)
):
    raise ValueError(
        "A coluna UF está desalinhada."
    )

if not rotuled["municipio"].reset_index(drop=True).equals(
    pph["MUNICIPIO"].reset_index(drop=True)
):
    raise ValueError(
        "A coluna MUNICIPIO está desalinhada."
    )

print("Validações de alinhamento concluídas com sucesso.")
```

```python
pph_selecionado = (
    pph[colunas_pph]
    .reset_index(drop=True)
    .copy()
)

rotuled_selecionado = (
    rotuled[colunas_rotuled]
    .reset_index(drop=True)
    .copy()
)

dataset_consolidado = pd.concat(
    [
        pph_selecionado,
        rotuled_selecionado
    ],
    axis=1
)

print("Dimensão do PPH selecionado:", pph_selecionado.shape)
print("Dimensão do rotuled selecionado:", rotuled_selecionado.shape)
print("Dimensão do dataset consolidado:", dataset_consolidado.shape)
```

```python
print("Quantidade total de colunas:", len(dataset_consolidado.columns))

print("\nPrimeiras 4 colunas:")
print(dataset_consolidado.columns[:4].tolist())

print("\nÚltimas 2 colunas:")
print(dataset_consolidado.columns[-2:].tolist())

print("\nExistem nomes de colunas duplicados?")
print(dataset_consolidado.columns.duplicated().any())
```

```python
print("Quantidade de linhas:", len(dataset_consolidado))

indice_esperado = pd.RangeIndex(
    start=0,
    stop=len(dataset_consolidado),
    step=1
)

print(
    "Índice sequencial de 0 até 27825:",
    dataset_consolidado.index.equals(indice_esperado)
)
```

```python
produtos_preservados = (
    dataset_consolidado["produtos_maior_consumo"]
    .equals(
        rotuled["produtos_maior_consumo"]
        .reset_index(drop=True)
    )
)

categorias_preservadas = (
    dataset_consolidado["categoria_maior_consumo"]
    .equals(
        rotuled["categoria_maior_consumo"]
        .reset_index(drop=True)
    )
)

print(
    "produtos_maior_consumo preservada:",
    produtos_preservados
)

print(
    "categoria_maior_consumo preservada:",
    categorias_preservadas
)
```

```python
valores_ausentes = (
    dataset_consolidado
    .isna()
    .sum()
    .sort_values(ascending=False)
)

colunas_com_ausentes = valores_ausentes[
    valores_ausentes > 0
]

print(
    "Quantidade de colunas com valores ausentes:",
    len(colunas_com_ausentes)
)

if len(colunas_com_ausentes) > 0:
    display(
        colunas_com_ausentes.to_frame(
            name="quantidade_ausentes"
        )
    )
else:
    print("Nenhum valor ausente encontrado.")
```

```python
CAMINHO_SAIDA = BASE_DIR / "training_dataset.csv"

print("Arquivo será salvo em:")
print(CAMINHO_SAIDA)
```

```python
dataset_consolidado.to_csv(
    CAMINHO_SAIDA,
    index=False
)

print(
    f"Dataset consolidado salvo com "
    f"{len(dataset_consolidado)} linhas e "
    f"{len(dataset_consolidado.columns)} colunas."
)

print(f"Caminho: {CAMINHO_SAIDA}")
```

```python
print(
    "Arquivo criado:",
    CAMINHO_SAIDA.exists()
)

if CAMINHO_SAIDA.exists():
    print(
        "Tamanho do arquivo em MB:",
        round(
            CAMINHO_SAIDA.stat().st_size / (1024 * 1024),
            2
        )
    )
```

```python
dataset_recarregado = pd.read_csv(CAMINHO_SAIDA)

print(
    "Dimensão do arquivo recarregado:",
    dataset_recarregado.shape
)

print(
    "Colunas preservadas:",
    dataset_recarregado.columns.tolist()
    == dataset_consolidado.columns.tolist()
)
```

```python
dados_preservados = dataset_recarregado.equals(
    dataset_consolidado
)

print(
    "Arquivo salvo preservou todos os dados:",
    dados_preservados
)
```

```python
from pandas.testing import assert_frame_equal

try:
    assert_frame_equal(
        dataset_recarregado,
        dataset_consolidado,
        check_dtype=False
    )

    print(
        "Arquivo salvo preservou os valores: True"
    )

except AssertionError as erro:
    print(
        "Arquivo salvo apresenta diferenças."
    )

    print(erro)
```
